```python
import pandas as pd
from nba_api.stats.static import players, teams
from nba_api.stats.endpoints import leaguegamefinder

player_id = players.find_players_by_full_name('Victor Wembanyama')[0]['id']   # 1641705
vs_player_id = players.find_players_by_full_name('Kevin Durant')[0]['id']    # 201142
team_id = teams.find_team_by_abbreviation('SAS')['id']                       # 1610612759

season = '2025-26'
season_id = '2' + season[:4]     # '22025', format a few endpoints use
season_year = int(season[:4])    # 2025
draft_year = 2023                # Wembanyama's draft, for the draft endpoints
```

```python
# the Spurs' last game of the season, for the single-game endpoints
games = leaguegamefinder.LeagueGameFinder(team_id_nullable=team_id, season_nullable=season).get_data_frames()[0]
games = games.sort_values('GAME_DATE', ascending=False)
game_id = games['GAME_ID'].iloc[0]
game_date = games['GAME_DATE'].iloc[0]
print(game_id, game_date, games['MATCHUP'].iloc[0])
```
